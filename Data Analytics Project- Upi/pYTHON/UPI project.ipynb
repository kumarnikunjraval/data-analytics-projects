### Step 2: Data Validation Using Excel 
### Ensure all customer_id values in upi_account_details, device_info, upi_transaction_history, and customer_feedback_surveys exist in customer_master.

```python

master_ids = set(customer_master["customer_id"])

def check_missing(df, table_name):
    missing = df.loc[~df["customer_id"].isin(master_ids), "customer_id"].unique()
    if len(missing) > 0:
        print(f"⚠️ Missing customer_ids in {table_name}: {missing}")
    else:
        print(f"✅ All customer_ids in {table_name} exist in customer_master.")

check_missing(upi_account_details, "upi_account_details")
check_missing(device_info, "device_info")
check_missing(upi_transaction_history, "upi_transaction_history")
check_missing(customer_feedback_surveys, "customer_feedback_surveys")
```

### Check that all merchant_id in upi_transaction_history exist in merchant_info.

```python
valid_merchants = set(merchant_info["merchant_id"])

missing_merchants = upi_transaction_history.loc[
    ~upi_transaction_history["merchant_id"].isin(valid_merchants),
    "merchant_id"
].unique()

if len(missing_merchants) > 0:
    print(f"⚠️ Missing merchant_ids in upi_transaction_history: {missing_merchants}")
else:
    print("✅ All merchant_ids in upi_transaction_history exist in merchant_info.")
```

### Validate that all upi_id in upi_transaction_history exist in upi_account_details.

```python
valid_upi_ids = set(upi_account_details["upi_id"])

missing_upi_ids = upi_transaction_history.loc[
    ~upi_transaction_history["upi_id"].isin(valid_upi_ids),
    "upi_id"
].unique()

if len(missing_upi_ids) > 0:
    print(f"⚠️ Missing upi_ids in upi_transaction_history: {missing_upi_ids}")
else:
    print("✅ All upi_ids in upi_transaction_history exist in upi_account_details.")
```

### Ensure all device_id in upi_transaction_history exist in device_info. 

```python
valid_device_ids = set(device_info["device_id"])

missing_device_ids = upi_transaction_history.loc[
    ~upi_transaction_history["device_id"].isin(valid_device_ids),
    "device_id"
].unique()

if len(missing_device_ids) > 0:
    print(f"⚠️ Missing device_ids in upi_transaction_history: {missing_device_ids}")
else:
    print("✅ All device_ids in upi_transaction_history exist in device_info.")
```


### All transaction_id in fraud_alert_history must exist in upi_transaction_history.

```python
valid_transaction_ids = set(upi_transaction_history["transaction_id"])

missing_transaction_ids = fraud_alert_history.loc[
    ~fraud_alert_history["transaction_id"].isin(valid_transaction_ids),
    "transaction_id"
].unique()

if len(missing_transaction_ids) > 0:
    print(f"⚠️ Missing transaction_ids in fraud_alert_history: {missing_transaction_ids}")
else:
    print("✅ All transaction_ids in fraud_alert_history exist in upi_transaction_history.")
```
